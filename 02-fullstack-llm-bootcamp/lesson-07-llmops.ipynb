## 1. Why LLMOps？

传统 ML 课程重点在于**如何构建模型**（训练神经网络、调超参等），但很少涵盖**如何将模型投入生产**。这一空白在 LLM 应用开发中尤为突出。

**LLMOps 的核心价值**：
- 覆盖 LLM 应用的完整生命周期：从模型选择 → 开发 → 测试 → 部署 → 监控 → 迭代
- 提供面向实际生产环境的工程实践框架
- 填补实验室原型与真实产品之间的工程 gap

> **关键认知**：构建一个可以运行的 LLM Demo 很容易，但将其打磨成可靠的生产系统需要系统化的工程方法。

## 2. 选择基础模型（Choosing Base Models）

选择基础模型需要在多个维度上进行权衡：

| 维度 | 说明 |
|------|------|
| **质量（Quality）** | 模型在目标任务上的表现 |
| **速度（Speed）** | 推理延迟，影响用户体验 |
| **成本（Cost）** | API 调用费用或自托管成本 |
| **可微调性（Tunability）** | 能否针对业务场景进行 fine-tuning |
| **数据安全（Data Security）** | 数据是否会被模型提供商用于训练 |

---

### 专有模型（Proprietary Models）

| 模型 | 特点 |
|------|------|
| **GPT-4** | 最高质量，适合复杂推理任务，成本较高 |
| **GPT-3.5-turbo** | 速度快、成本低，适合大多数场景 |
| **Claude** | Anthropic 出品，在安全性和长上下文处理上有优势 |
| **Cohere** | 适合企业级 NLP 任务，提供 fine-tuning 支持 |

### 开源模型（Open-source Models）

| 模型 | 说明 |
|------|------|
| **T5 / Flan-T5** | Google 出品，适合结构化 NLP 任务，支持 fine-tuning |
| **LLaMA** | Meta 出品，研究许可，社区生态丰富 |
| **Alpaca** | 基于 LLaMA 指令微调，Stanford 出品 |
| **Vicuna** | 基于 LLaMA，通过 ChatGPT 对话数据微调 |

> **注意**：开源模型的许可证各不相同，商业用途前务必确认许可证类型。Apache 2.0 许可证对商业友好，LLaMA 原始许可证限制商业使用。

### 推荐策略

- **起点**：GPT-4 是大多数新项目的好起点——质量高，方便快速验证想法
- **降本**：验证可行性后，可以尝试用 GPT-3.5 或开源模型替代，以降低成本
- **数据敏感场景**：考虑自托管开源模型，避免数据外泄

## 3. Prompt 管理（Prompt Management）

### 当前痛点

LLM 应用开发中，prompt 的迭代是核心工作，但目前业界**缺乏成熟的工程化实践**：
- Prompt 散落在代码各处，难以追踪版本
- 多人协作时容易产生冲突
- 缺少系统化的 prompt 实验管理工具

### 三级 Prompt 追踪方案

根据团队规模和需求，可以选择不同层级的方案：

**Level 1：OpenAI Playground（原型阶段）**
- 适合快速原型验证
- 缺点：无版本控制，难以与代码集成

**Level 2：Git 版本控制（推荐默认方案）**
- 将 prompt 作为代码的一部分纳入 Git 管理
- 优点：零额外成本，与现有工程流程无缝集成
- 适合大多数团队的起点

**Level 3：专业 Prompt 追踪工具**
- 适合需要与非技术成员（产品、运营）协作的场景
- 提供可视化界面和 A/B 测试能力

### 新兴 Prompt 管理工具

| 工具 | 定位 |
|------|------|
| **Weights & Biases** | 原 ML 实验追踪工具，扩展了 LLM prompt 管理能力 |
| **Comet** | 类似 W&B，支持 prompt 版本和实验对比 |
| **MLflow** | 开源 ML 生命周期管理，社区在探索 LLM 支持 |

> 这一领域工具正在快速发展，选择时优先考虑与现有工具链的兼容性。

## 4. 测试与评估（Testing & Evaluation）

### 三大核心挑战

1. **训练数据未知**：API 模型（如 GPT-4）的训练数据不透明，难以预判模型的知识边界
2. **分布偏移（Distribution Shift）**：生产环境中用户的真实输入与开发阶段的测试数据往往有差距
3. **评估指标不明确**：很多 LLM 任务（如摘要、对话）难以用单一数字量化质量

### 构建评估集的最佳实践

- **增量构建**：不要等到开发完成再建评估集，在开发过程中持续积累
- **覆盖边缘 case**：包含困难样本、罕见输入、以及容易出错的类型
- **用 LLM 辅助生成测试数据**：利用模型生成多样化的输入变体，扩展覆盖范围
- **对齐用户任务分布**：评估集应尽量反映真实用户的使用场景

> **核心原则**：要确认对模型或 prompt 的修改是否有效，必须在能代表真实用户输入的宽泛数据集上进行测量。

### 评估指标选择策略

```
有明确正确答案？
  ├─ 是 → 准确率（Accuracy）
  └─ 否 → 有参考答案？
              ├─ 是 → 语义相似度（Semantic Similarity）
              └─ 否 → LLM 评判（LLM-as-Judge）或 人工反馈
```

| 指标 | 适用场景 | 示例 |
|------|----------|------|
| **Accuracy** | 分类、问答等有标准答案的任务 | 情感分析、选择题 |
| **Semantic Similarity** | 有参考答案但答案形式灵活 | 翻译质量、摘要对比 |
| **LLM-as-Judge** | 开放式生成，无标准答案 | 对话质量、创意写作 |
| **Human Feedback** | 最可靠，但成本高 | 最终质量验收 |

## 5. 部署与监控（Deployment & Monitoring）

### 提升输出质量的技术

**Self-critique（自我审查）**
- 让模型生成答案后，再对自己的输出进行评审和修正
- 本质上是 prompt 链：生成 → 批判 → 修订

**多次采样（Multiple Sampling）**
- 对同一输入生成多个候选输出
- 通过选择最优输出（可用另一个 LLM 评判）提升质量
- 代价：增加 API 调用次数，成本上升

**集成方法（Ensemble）**
- 结合多个模型的输出来提高鲁棒性
- 类似传统 ML 中的 ensemble 策略

---

### 监控策略

**核心监控目标：用户满意度**

- **低摩擦反馈机制**：在产品中加入简单的 👍 / 👎 按钮，降低用户反馈成本
- **自动化指标**：监控响应长度、错误率、超时率、拒绝率等代理指标
- **日志记录**：记录完整的输入/输出，供后续分析和测试集构建使用

| 监控维度 | 指标示例 |
|----------|----------|
| **用户满意度** | 👍/👎 比率、会话完成率 |
| **输出质量** | 响应长度分布、空响应率 |
| **系统性能** | 延迟（P50/P95/P99）、错误率 |
| **成本** | Token 使用量、API 调用次数 |

## 6. 持续改进循环（Continuous Improvement Cycle）

LLMOps 的核心理念是**测试驱动开发（Test-Driven Development）**与**持续反馈循环**的结合。

### 改进循环流程

```
┌─────────────────────────────────────────────────────┐
│                                                     │
│   开发（Development）                               │
│       ↓                                             │
│   部署（Deployment）                                │
│       ↓                                             │
│   收集用户反馈（User Feedback）                     │
│       ↓                                             │
│   日志分析 & 监控（Logging & Monitoring）           │
│       ↓                                             │
│   提取失败案例 → 扩充测试集                         │
│       ↓                                             │
│   迭代优化 Prompt / 微调模型                        │
│       ↓                                             │
│   回到「开发」↑                                     │
└─────────────────────────────────────────────────────┘
```

### 关键实践原则

1. **从失败中学习**：将生产环境中的失败案例直接转化为测试集，驱动下一轮改进
2. **数据飞轮**：用户越多 → 反馈越丰富 → 测试集越完善 → 模型越可靠 → 用户越满意
3. **小步快跑**：相比一次性大优化，更推荐高频的小迭代
4. **先测试再改动**：每次修改 prompt 或切换模型前，先确保有覆盖该功能的测试用例

> **总结**：LLMOps 不是一套一次性的设置，而是一个持续运转的工程体系。它的目标是让 LLM 应用像互联网产品一样，能够基于真实用户反馈快速迭代和演进。
